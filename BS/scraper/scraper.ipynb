```python
url = 'https://www.imdb.com'
sub_url = '/search/title/?groups=top_250&sort=user_rating,desc&start=1&ref_=adv_nxt'
page = get(url+sub_url)
page_soup = BeautifulSoup(page.text, 'html.parser')
top_movie = page_soup.find_all('div', class_='lister-item-content')
movie_cast = {}
actor_movies = {}
for movie in top_movie:
    movie_name = movie.h3.a.text
    print(f'Movie name - {movie_name}')
    movie_url = url + movie.h3.a['href']
    movie_page = get(movie_url)
    movie_soup = BeautifulSoup(movie_page.text, 'html.parser')
    cast_list = movie_soup.find_all('div', 'StyledComponents__CastItemSummary-y9ygcu-9 fBAofn')
    movie_cast[movie_name] = []
    for cast in cast_list:
        actor = cast.a.text
        movie_cast[movie_name].append(actor)
        if actor in actor_movies.keys():
            actor_movies[actor].append(movie_name)
        else:
            actor_movies[actor] = [movie_name]
```

```python
name = input('Actor name:')
n = input('Number of movies')
name_list = actor_movies[name]
for i in range(0, int(n)):
    print(name_list[i])
```
